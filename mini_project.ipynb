# Practice Question — Topic 8
Build an inventory system for a small shop:

Create a list of 5 products where each product is a tuple: (name, price, quantity)
Add 2 more products using .append()
Remove one product by unpacking its name and using .remove()
Iterate using enumerate() to display a numbered inventory
Calculate: total inventory value (price × quantity for all), most expensive item

```python
# PRACTICE QUESTION — Topic 8
# Write your inventory system here
# Create list of product with tuple: (name, price, quantity)
products = [ ("Golden Penny Spaghetti", 1100, 20), 
            ("Indomie Instant Noodles", 300, 24), 
            ("Chikki Instant Noodles", 300, 40), 
            ("Golden Penny Semovita", 1800, 40), 
            ("Auntie B Spaghetti", 900, 20), 
           
]

print("=" * 70)
print("          PRODUCT INVENTORY SYSTEM")
print("=" * 70)

print("PRODUCT NAME:")
for product in products:
    print(product)

# Adding two more products using append
products.append(("Maltina", 7200, 12)),
products.append(("Pepsi", 6000, 12)), 

print("-" * 70)

print("PRODUCT NAME AFTER APPEND:")
for product in products:
    print(product)

print("-" * 70)

# Removing one product by unpacking its name
name, price, quantity = products[0]
products.remove((name, price, quantity))

print("PRODUCT NAME AFTER REMOVING:")
for product in products:
    print(product)

print("-" * 70)

# Calculate Total Inventory Value
total_value = 0

for name, price, quantity in products:
    total_value += price * quantity

print("TOTAL INVENTORY VALUE =", total_value)
      
print("-" * 70)

# Calculate most expensive item
most_expensive = products[0]

for product in products:
    if most_expensive[1] > product[1]:
        most_expensive = product 

name, price, quantity = most_expensive

print("MOST EXPENSIVE ITEM:", name)
print("PRICE:", price)
print("QUANTITY:", quantity)

print("=" * 70)
```

## Practice Question — Topic 9

Build a **Course Enrollment Tracker**:

1. Create a dictionary mapping course names to a set of enrolled students
2. Add at least 2 more students to one course using `.add()` on the set
3. Use set operations to find: students in both Course A and Course B (intersection), students in Course A only (difference)
4. Iterate the dictionary with `.items()` to print each course and student count
5. Find which course has the most students

```python
# PRACTICE QUESTION — Topic 9
# Write your Course Enrollment Tracker here
print("=" * 95)
print("        COURSE ENROLLMENT TRACKER")
print("=" * 95)

course = {
    "Data Science": {"Mercy", "Hadiyah", "Taiwo"},
    "Web Development": {"Enola", "Williams", "Alice"},
    "Cybersecurity": {"Louise", "Don", "Lilly", "Blossom"}
}

# Adding two student name in one course

course["Data Science"].add("Christabell")
course["Data Science"].add("Hameedat Blue")

print(course)

print("-" * 95)

# To find intersection between course A and B & Students in course A only difference
print(course["Data Science"].intersection(course["Web Development"]))

print( )

print(course["Data Science"].difference(course["Web Development"]))
print("." * 95)

print(course["Web Development"] & course["Cybersecurity"])

print( )

print(course["Web Development"] - course["Cybersecurity"])

print("-" * 95)

for course_name, students in course.items():
    print(course_name)
    print(students)
    print()

print("-" * 95)

for course_name, students in course.items():
    print(f"course: {course_name}")
    print("Students:")

    for student in students:
        print(f"- {student}")
        
    print("-" * 95)

# To find the course that has the most student
most_students = ""
highest_count = 0

for course_name, students in course.items():
    if len(students) > highest_count:
        highest_count = len(students)
        most_students = course_name

print("Course with the Most Students:", most_students)
print("Number of most students:", highest_count)

print("=" * 95)
```
