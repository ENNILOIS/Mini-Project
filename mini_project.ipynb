#### MINI PROJECT

Goal: Build a simple Python "Identity Card" generator.

Requirements:
  1. Ask the user for: full name, age, country, favourite Python use case, and one goal for this course
  2. Display a formatted identity card using print() statements and =/- borders
  3. Include a motivational footer message



```python
# Display Python identity card border
print("=" * 66)
print("                PYTHON LEARNER IDENTITY CARD")
print("=" * 66)

# Ask the user for their full name
Full_name = input("Please input your full name: ")

# Ask the user for their age
Age = input("Please input your age: ")

# Ask the user for their country
Country = input("Please input your country: ")

# Ask the user for their favourite python use case
Python_use = input("Please input your python use case: ")

# Ask the user for their goal for the course
Course_goal = input("Please input the goal for this course")

# Display identity card message
print("Name    :",Full_name)
print("Age     :",Age)
print("Country :",Country)
print("Focus   :",Python_use)
print("Goal    :",Course_goal)
print("-" * 66)
print(' "Every expert was once a beginner. Keep learning, keep growing." ')
print("=" * 66)
```
