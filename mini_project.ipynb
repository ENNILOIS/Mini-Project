# Practice Question — Topic 8
Build an inventory system for a small shop:

Create a list of 5 products where each product is a tuple: (name, price, quantity)
Add 2 more products using .append()
Remove one product by unpacking its name and using .remove()
Iterate using enumerate() to display a numbered inventory
Calculate: total inventory value (price × quantity for all), most expensive item

```python
# PRACTICE QUESTION — Topic 8
# Write your inventory system here
# Create list of product with tuple: (name, price, quantity)
products = [ ("Golden Penny Spaghetti", 1100, 20), 
            ("Indomie Instant Noodles", 300, 24), 
            ("Chikki Instant Noodles", 300, 40), 
            ("Golden Penny Semovita", 1800, 40), 
            ("Auntie B Spaghetti", 900, 20), 
           
]

print("=" * 70)
print("          PRODUCT INVENTORY SYSTEM")
print("=" * 70)

print("PRODUCT NAME:")
for product in products:
    print(product)

# Adding two more products using append
products.append(("Maltina", 7200, 12)),
products.append(("Pepsi", 6000, 12)), 

print("-" * 70)

print("PRODUCT NAME AFTER APPEND:")
for product in products:
    print(product)

print("-" * 70)

# Removing one product by unpacking its name
name, price, quantity = products[0]
products.remove((name, price, quantity))

print("PRODUCT NAME AFTER REMOVING:")
for product in products:
    print(product)

print("-" * 70)

# Print Enumerated Entry
for num, product in  enumerate(products, start=1):
    print(f"{num} - {product}")
# if you want only the product name 
  #  print(f"{num} - {product[0]}")

print("-" * 70)

# Calculate Total Inventory Value
total_value = 0

for name, price, quantity in products:
    total_value += price * quantity

print("TOTAL INVENTORY VALUE =", total_value)
      
print("-" * 70)

# Calculate most expensive item
most_expensive = products[0]

for product in products:
    if most_expensive[1] > product[1]:
        most_expensive = product 

name, price, quantity = most_expensive

print("MOST EXPENSIVE ITEM:", name)
print("PRICE:", price)
print("QUANTITY:", quantity)

print("=" * 70)
```

## Practice Question — Topic 10

Build a **Student Grade Calculator** using functions:

1. Define `get_scores()` — asks the user to enter 3 exam scores and returns them as a tuple
2. Define `calculate_average(scores)` — takes the tuple and returns the average
3. Define `classify_grade(average)` — returns the letter grade (A ≥ 80, B ≥ 70, C ≥ 60, D ≥ 50, else F)
4. Define `display_report(name, scores, average, grade)` — prints a formatted report
5. In the main program: ask the user's name, call each function in order, and display the report

**Bonus:** Add a default parameter `passing_score=50` to `classify_grade()` so the pass threshold is adjustable.


```python
# PRACTICE QUESTION — Topic 10
# Write your Student Grade Calculator here

print("=" * 55)
print(f"          BUILT STUDENT GRADE CALCULATOR")
print("=" * 55)

def users_name():
    name = input("Users enter your full name: ")
    return name

name = users_name()

def get_scores():
    score1 = int(input(" Users should enter their first exam score: "))
    score2 = int(input("Users should enter their second exam score: "))
    score3 = int(input("Users should enter their third exam score:  "))
    return (score1, score2, score3)

scores = get_scores()

def calculate_average(scores):
    return sum(scores) / len(scores)

average = calculate_average(scores)

def classify_grade(average_score, passing_score=50):
    if average_score >= 80:
        return "A"
    elif average_score >= 70:
        return "B"
    elif average_score >= 60:
        return "C"
    elif average_score >= 50:
        return "D"
    else:
        return "F"
        
grade = classify_grade(average)

def display_report(name, scores, average, grade):
    print("-" * 55)
    print(f"\n        USERS REPORT RECORD:")
    print("-" * 55)
    print("\nFULL NAME:", name)
    print("\nSCORES:", scores)
    print("\nAVERAGE:", average)
    print("\nGRADE:", grade)

display_report(name, scores, average, grade)

print("=" * 55)
```
