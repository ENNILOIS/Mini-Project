```python
import csv
import numpy as np

transactions = []

with open("transactions.csv", encoding="utf-8-sig")as f:
    reader = csv.DictReader(f)
    
    for row in reader:
        transactions.append(row)

balance_after = np.array([row["balance_after"] for row in transactions 
                          if row["balance_after"].strip() !=""], dtype=float)

print(balance_after)
```

```python
balances = balance_after

mean_val = balances.mean()
median_val = np.median(balances)

print("==== Central Tendency: balance_after ====")
print(f"Mean:   NGN {mean_val:>10,.2f}")
print(f"Median: NGN {median_val:>10,.2f}")
print()

# Notifying the Gap between the Mean & the Median and what it signals about the data
print(f"Mean - Median Gap:")
print(f"NGN {mean_val - median_val:>10,.2f}")
print(f"DIRECTION: \n  Mean > Median → SLIGHTLY RIGHT-SKEWED (Positive Skewed) distribution")
print(f"INTERPRETATION: \n   The small mean-median gap of ₦1,498.01 suggests that the distribution is only midly right-skewed,\n with a few higher balance values pulling the mean slightly above the median.")
     
```

### QUALITY PROBLEMS IDENTIFIED FROM THE DATASET OUTPUTS

1. There are missing values in the **phone** column (8 Missing) and in the **balance_after** column (6 Missing) that need to be treated.

2. The **transaction_amount** is in the wrong datatype format; it should be a calculated number (float) not to be treated as a string, because of currency symbols like "₦".

3. Also, the **transaction_date** is in the wrong datatype format too and should be treated as a datetime, not a string.

4. In the **branch_state** column, there is inconsistency in their naming and capitalization (such as "Kaduna" vs. "KADUNA").

5. The **phone** column is in the wrong datatype format as well; it should be stored as a string instead of a float so it does not display in scientific notation.

##### CLEANING THE DATA

```python
# WORKING ON A COPY

df_clean = df.copy()

print("Working Copy Created:", df_clean.shape)
print("Original Dataset:    ", df.shape)
```

```python
# Dealing with Missing Values by filling with Not Provided

df_clean["phone"] = df_clean["phone"].fillna("Not Provided")

print("Missing Phone Values Filled:", df_clean["phone"].isnull().sum())
print("Sample phone values after filling missing values:")
print(df_clean["phone"].value_counts().head())
```

```python
# Changing the Phone Column to String Type

df["phone"] = df["phone"].astype("Int64").astype(str)

df.head(3)
```

### **Method Specified:**
Impute missing phone numbers with the placeholder string 'Not Provided'.

### **Technical Justification:**
Phone numbers represent unique Personal Identifiable Information (PII) that cannot be accurately estimated or imputed statistically. Dropping these rows would result in unnecessary data loss for non-contact-related analysis. Using 'Not Provided' preserves the transaction record while maintaining data integrity.

```python
# Dropping rows with missing values in the 'balance_after' column

before_drop_shape = df_clean.shape
df_clean = df_clean.dropna(subset=["balance_after"])
after_drop_shape = df_clean.shape

print(f"Before Drop Null Values: {before_drop_shape[0]} Rows")
print(f"After Drop Null Values:   {after_drop_shape[0]} Rows")
print(f"Null Values Dropped:        {before_drop_shape[0] - after_drop_shape[0]}")
print(f"Missing balance_after Values: {df_clean['balance_after'].isnull().sum()}")
```

### **Method Specified:**
Drop rows containing missing values in the balance_assets column.

### **Technical Justification:**
Asset balances are strict financial metrics that should not be estimated using statistical imputation (e.g., mean or median), as this introduces artificial variance into financial calculations. Dropping these records ensures that balance aggregations reflect only true, verified balances.

```python
# TO CHECK DUPLICATE ROWS IN THE DATASET

print(f"Duplicate Rows: {df_clean.duplicated().sum()}")
print()
print("THE DUPLICATE ROWS:")
df[df.duplicated(keep=False)].sort_values("balance_after")
```

```python
# To Remove Duplicate Rows in the Dataset

before_drop_shape = df_clean.shape
df_clean = df_clean.drop_duplicates()
after_drop_shape = df_clean.shape

print(f"Before Drop Duplicates: {before_drop_shape[0]} Rows")
print(f"After Drop Duplicates:   {after_drop_shape[0]} Rows")   
print(f"Duplicate Rows Dropped:  {before_drop_shape[0] - after_drop_shape[0]}")
```

### **Method Specified:**
Identify and drop duplicate rows across the dataset.

### **Technical Justification:**
Duplicate entries artificially inflate transaction volumes and distort summary metrics. Removing redundant rows ensures that each transaction is uniquely counted, guaranteeing accurate downstream aggregations.

```python
df_clean.info()
```

```python
# TO STRIP THE ₦ SYMBOL FROM THE TRANSACTION_AMOUNT COLUMNS AND CONVERT TO FLOAT.
print("Before — dtype:", df_clean["transaction_amount"].dtype)
print("Before — sample:", df_clean["transaction_amount"].head(5).tolist())

# Step 1: remove the ₦ symbol
df_clean["transaction_amount"] = df_clean["transaction_amount"].str.replace("₦", "", regex=False)

# Step 2: convert to numeric (errors='coerce')
df_clean["transaction_amount"] = pd.to_numeric(df_clean["transaction_amount"], errors='coerce')

print()
print("After — dtype:", df_clean["transaction_amount"].dtype)
print("After — sample:", df_clean["transaction_amount"].head(5).tolist())
print("Find Any Missing New Values?", df_clean["transaction_amount"].isnull().sum())
```

### **Method Specified:**
Strip the currency symbol (₦) and commas, then convert the column data type from string to numeric (float/int).

### **Technical Justification:**
Financial values formatted as strings with currency symbols block mathematical operations and statistical aggregation. Removing non-numeric characters and casting the column to a numeric data type enables numerical computations, KPI calculations, and plotting.

```python
# TO STANDARDISE BRANCH_STATE TO TITLE CASE
print("Before - unique branch_state value:")
print(sorted(df_clean["branch_state"].unique()))
print(f"Total Unique Values: {df_clean['branch_state'].nunique()}")

df_clean["branch_state"] = df_clean["branch_state"].str.strip().str.title()

print()
print()
print("After - unique branch_state value:")
print(sorted(df_clean["branch_state"].unique()))
print(f"Total Unique Values: {df_clean['branch_state'].nunique()}")
```

### **Method Specified:**
Trim leading/trailing whitespace and apply Title Case formatting to the branch names.

### **Technical Justification:**
Variations in casing (e.g., "ANAMBRA" vs "anambra") and hidden spaces cause data processing engines to treat identical locations as separate categories. Standardizing string formatting guarantees accurate categorical grouping during GROUP BY operations and visualization.

```python
# TO CONVERT TRANSACTION_DATE TO DATETIME 
print("Before — dtype:", df_clean["transaction_date"].dtype)
print("Before — sample:", df_clean["transaction_date"].head(3).tolist())

df_clean["transaction_date"] = pd.to_datetime(df_clean["transaction_date"])

print()
print()
print("After — dtype:", df_clean["transaction_date"].dtype)
print("After — sample:", df_clean["transaction_date"].head(3).tolist())
```

```python
# TO EXTRACT MONTH COLUMN FROM TRANSACTION_DATE

df_clean["transaction_month"] = df_clean["transaction_date"].dt.month

print("Transaction Per Month:")
print(df_clean["transaction_month"].value_counts().sort_index())
```

### **Method Specified:**
Convert transaction_date to datetime format and extract the month into a separate column.

### **Technical Justification:**
String-based dates cannot be sorted sequentially or grouped by time intervals. Parsing to datetime enables time-series tracking and allows the extraction of time attributes (such as month), which supports temporal trends analysis and dashboard filtering.

##### TRANSFORM & ANSWER BUSINESS QUESTIONS

```python
# USING GROUPBY.AGG TO ANSWER THE QUESTIONS

# Step 1: Group by state and calculate the three metrics
state_summary = df_clean.groupby("branch_state").agg(
    total_volume=("transaction_amount", "sum"),
    order_count=("transaction_amount", "count"),
    avg_transaction=("transaction_amount", "mean")
)
```

```python
# Question 1: Branch State Summary Table
state_summary = df_clean.groupby("branch_state").agg(
    total_volume=("transaction_amount", "sum"),
    order_count=("transaction_amount", "count"),
    avg_transaction=("transaction_amount", "mean")
).sort_values(by="total_volume", ascending=False)

# Display the summary table
state_summary.head(10)
```

##### INTERPRETING THE BRANCH STATE ANALYSIS FOR BUSINESS
1. Top Revenue Generator: Ogun State ranks highest in total transaction volume at $\text{₦}51,178,475.79$ across 57 transactions, achieving the highest average order value ($\text{₦}897,867.89$) among all states.

2. Volume Leaders: Delta ($\text{₦}48.77\text{M}$) and Kaduna ($\text{₦}30.95\text{M}$) follow as key volume drivers.

3. Business Impact: Ogun and Delta are high-value markets for the bank. Operations should prioritize credit line availability and high-tier customer support services in these branches to retain high-value account holders.

```python
# Question 2: Channel Summary Table
channel_summary = df_clean.groupby("channel").agg(
    frequency=("transaction_amount", "count"),
    avg_transaction=("transaction_amount", "mean")
).sort_values(by="frequency", ascending=False)

# Display the summary table
channel_summary.head(10)
```

##### INTERPRETING THE CHANNEL PERFORMANCE ANALYSIS FOR BUSINESS
1. Dominant Channel: POS is both the most frequently used channel (110 transactions) and yields the highest average transaction amount ($\text{₦}645,699.25$).

2. Secondary Channels: Bank Branch remains popular for high-value operations (103 transactions, averaging $\text{₦}343,017.75$), while Mobile App handles smaller, lower-value transactions (80 transactions, averaging $\text{₦}391,241.11$).

3. Business Impact: Since POS handles the largest throughput, technical infrastructure for POS transaction processing must remain reliable to prevent revenue loss. The bank can also run campaigns to convert Mobile App users into higher-volume transactions.
